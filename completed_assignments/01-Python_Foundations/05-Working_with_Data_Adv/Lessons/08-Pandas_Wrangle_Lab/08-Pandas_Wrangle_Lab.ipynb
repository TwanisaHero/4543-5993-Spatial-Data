### 📥 1. Load the Dataset

```python
import pandas as pd
from pathlib import Path

student_path = Path("../../data/students.csv")
if not student_path.exists():
    sample_students = pd.DataFrame({
        "first_name": ["Ava", "Ben", "Cora", "Diego", "Ella", "Finn"],
        "last_name": ["Nguyen", "Smith", "Patel", "Garcia", "Brown", "Jones"],
        "math_score": [95, 82, None, 74, 68, 91],
        "science_score": [88, 79, 93, None, 72, 96],
    })
    student_path.parent.mkdir(parents=True, exist_ok=True)
    sample_students.to_csv(student_path, index=False)

df = pd.read_csv(student_path)
df.head()
```

### 🔎 2. Inspect and Audit the Data

```python
# Basic overview
df.info()
```

```python
# Summary stats
df.describe()
```

### 🧼 3. Clean Missing or Invalid Data

```python
# Check for missing values
df.isnull().sum()
```

```python
# Drop rows with missing names
df = df.dropna(subset=["first_name", "last_name"]).copy()

# Fill any missing scores with column averages
df["math_score"] = df["math_score"].fillna(df["math_score"].mean())
df["science_score"] = df["science_score"].fillna(df["science_score"].mean())
```

### 🧠 4. Feature Engineering (New Columns)

```python
# Add average and grade
df["average_score"] = (df["math_score"] + df["science_score"]) / 2

def grade(score):
    if score >= 90:
        return "A"
    elif score >= 80:
        return "B"
    elif score >= 70:
        return "C"
    else:
        return "D"

df["grade"] = df["average_score"].apply(grade)
df.head()
```

### 🔽 5. Sorting and Filtering

```python
# Top performers
df[df["average_score"] > 90].sort_values(by="average_score", ascending=False).head()
```

### 📊 6. Group and Describe by Grade

```python
# How many of each grade?
df["grade"].value_counts()
```

```python
# Average scores per grade group
df.groupby("grade")[["math_score", "science_score", "average_score"]].mean()
```

### 💾 7. Save the Cleaned Dataset

```python
df.to_csv("student_scores_cleaned.csv", index=False)
```
