## 1. Create a small dataset and compute summary statistics.

```python
# Answer 1
import pandas as pd

df = pd.DataFrame({
    "city": ["A", "B", "C", "D", "E"],
    "temperature": [71, 75, 78, 72, 85],
    "rain_inches": [1.2, 0.5, 0.0, 2.1, 0.3],
})
df.select_dtypes("number").describe()
```

## 2. Make a bar chart.

```python
# Answer 2
import matplotlib.pyplot as plt

plt.bar(df["city"], df["temperature"])
plt.xlabel("City")
plt.ylabel("Temperature")
plt.title("Temperature by City")
plt.show()
```

## 3. Make a scatterplot and describe the relationship.

```python
# Answer 3
plt.scatter(df["temperature"], df["rain_inches"])
plt.xlabel("Temperature")
plt.ylabel("Rain inches")
plt.title("Temperature vs Rain")
plt.show()

print("In this small sample, higher temperature does not appear to strongly predict more rain.")
```

## 4. Identify an outlier.

```python
# Answer 4
outlier = df.loc[df["temperature"].idxmax()]
print(outlier)
```
