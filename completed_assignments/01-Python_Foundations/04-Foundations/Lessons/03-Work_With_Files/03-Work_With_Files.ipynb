## 2. Writing and Reading Text Files

```python
# Writing to a text file
with open("example.txt", "w") as f:
    f.write("Hello, Jupyter!\n")
    f.write("This is a text file.\n")
```

```python
# Reading from a text file
with open("example.txt", "r") as f:
    contents = f.read()
contents
```

✅ **Your Turn**: Write your name into a file called `me.txt`, then read it back and print the result.

```python
# Completed reference solution: me file
with open("me.txt", "w", encoding="utf-8") as file:
    file.write("Demii")
with open("me.txt", "r", encoding="utf-8") as file:
    print(file.read())
```

## 3. JSON Files

```python
import json

data = {"name": "Graylian", "type": "mascot", "skills": ["howl", "parrot talk"]}

# Write JSON
with open("data.json", "w") as f:
    json.dump(data, f)

# Read JSON
with open("data.json", "r") as f:
    loaded = json.load(f)
loaded
```

✅ **Your Turn**: Create a dictionary with your name and 2-3 favorite courses. Save it to `courses.json`, then read it back.

```python
# Completed reference solution: courses json
import json
courses = {"name": "Demii", "favorite_courses": ["Spatial Data", "Python", "Data Science"]}
with open("courses.json", "w", encoding="utf-8") as file:
    json.dump(courses, file, indent=2)
with open("courses.json", "r", encoding="utf-8") as file:
    print(json.load(file))
```

## 4. CSV Files

```python
import csv

# Writing CSV
rows = [["Name", "Score"], ["Alice", 90], ["Bob", 85]]
with open("scores.csv", "w", newline="") as f:
    writer = csv.writer(f)
    writer.writerows(rows)

# Reading CSV
with open("scores.csv", "r") as f:
    reader = csv.reader(f)
    for row in reader:
        print(row)
```

✅ **Your Turn**: Create a CSV with 3 rows of your own data (e.g., pets, grades, or hobbies). Read it back and display the rows.

```python
# Completed reference solution: hobbies csv
import csv
rows = [["name", "hobby", "hours_per_week"], ["Ava", "reading", 4], ["Ben", "gaming", 6], ["Cora", "running", 3]]
with open("hobbies.csv", "w", newline="", encoding="utf-8") as file:
    csv.writer(file).writerows(rows)
with open("hobbies.csv", "r", encoding="utf-8") as file:
    for row in csv.reader(file):
        print(row)
```

---
### Summary
- `%pwd`, `%ls`, `%cd` show you where you are in the filesystem.
- Use `with open(...)` to safely read/write files.
- JSON is for structured data (key/value).
- CSV is for tabular data (rows/columns).
- Managing files is step 1 in any real data science project.
