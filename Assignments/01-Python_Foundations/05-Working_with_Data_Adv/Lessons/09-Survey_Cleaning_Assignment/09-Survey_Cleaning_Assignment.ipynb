```python
import pandas as pd
import numpy as np

# The repo includes survey.csv rather than survey_dirty.csv; this file is intentionally messy.
df = pd.read_csv("../../data/survey.csv")
df.head()
```

```python
# 1. Load and inspect the data
print(df.shape)
display(df.head())
df.info()
```

```python
# 2. Clean up string fields
for col in df.select_dtypes(include="object").columns:
    df[col] = df[col].astype("string").str.strip()
    df[col] = df[col].replace({"": pd.NA, "nan": pd.NA, "NaN": pd.NA})

df.head()
```

```python
# 3. Handle missing values
missing_before = df.isna().sum().sort_values(ascending=False)
missing_before
```

```python
# 4. Standardize categorical data
gender_map = {
    "m": "Male", "male": "Male", "man": "Male",
    "f": "Female", "female": "Female", "woman": "Female",
    "other": "Other", "nonbinary": "Other", "non-binary": "Other",
}
df["gender"] = (
    df["gender"].astype("string").str.strip().str.lower().map(gender_map).fillna("Other")
)

major_map = {
    "compsci": "Computer Science",
    "computer science": "Computer Science",
    "cs": "Computer Science",
    "data science": "Data Science",
    "math": "Mathematics",
    "mathematics": "Mathematics",
}
df["major"] = df["major"].astype("string").str.strip().str.lower().map(major_map).fillna(df["major"])

df[["gender", "major"]].head()
```

```python
# 5. Fix data types
def clean_number(series):
    return pd.to_numeric(series.astype("string").str.extract(r"(-?\d+(?:\.\d+)?)")[0], errors="coerce")

for col in ["age", "gpa", "study_hours_per_week", "satisfaction_1_5", "expected_grad_year", "commute_minutes"]:
    df[col] = clean_number(df[col])

df["survey_date"] = pd.to_datetime(df["survey_date"], errors="coerce")
df["attendance_rate"] = clean_number(df["attendance_rate"])

df.dtypes
```

```python
# 6. Validate and clean bounded numeric values
df.loc[(df["age"] < 16) | (df["age"] > 80), "age"] = np.nan
df.loc[(df["gpa"] < 0) | (df["gpa"] > 4), "gpa"] = np.nan
df.loc[(df["study_hours_per_week"] < 0) | (df["study_hours_per_week"] > 100), "study_hours_per_week"] = np.nan
df.loc[(df["satisfaction_1_5"] < 1) | (df["satisfaction_1_5"] > 5), "satisfaction_1_5"] = np.nan
df.loc[(df["attendance_rate"] < 0) | (df["attendance_rate"] > 100), "attendance_rate"] = np.nan

df.select_dtypes("number").describe()
```

```python
# 7. Validate and clean emails if an email column exists
if "email" in df.columns:
    valid_email = df["email"].astype("string").str.match(r"^[^@\s]+@[^@\s]+\.[^@\s]+$", na=False)
    df.loc[~valid_email, "email"] = pd.NA
else:
    print("No email column exists in this survey file.")
```

```python
# 8. Convert yes/no style fields to booleans
def to_bool(value):
    if pd.isna(value):
        return pd.NA
    text = str(value).strip().lower()
    if text in {"yes", "y", "true", "1"}:
        return True
    if text in {"no", "n", "false", "0"}:
        return False
    return pd.NA

for col in ["dorm_resident", "clubs_member"]:
    if col in df.columns:
        df[col] = df[col].apply(to_bool).astype("boolean")

df[["dorm_resident", "clubs_member"]].head()
```

```python
# 9. Check for duplicates
duplicate_count = df.duplicated().sum()
print("Duplicate rows:", duplicate_count)
df = df.drop_duplicates().copy()
```

```python
# 10. Fill or flag remaining missing values
numeric_cols = df.select_dtypes(include="number").columns
for col in numeric_cols:
    df[col] = df[col].fillna(df[col].median())

for col in df.select_dtypes(include="string").columns:
    df[col] = df[col].fillna("Unknown")

cleaned_missing = df.isna().sum().sort_values(ascending=False)
cleaned_missing
```

```python
# 11. Save the cleaned file
df.to_csv("survey_cleaned.csv", index=False)
df.head()
```

## Cleaning Summary

The survey file was cleaned by trimming string columns, standardizing categories, coercing numeric fields, removing impossible values, converting yes/no fields into booleans, dropping duplicates, filling remaining missing values, and saving the result as `survey_cleaned.csv`. Some fields, such as invalid dates and out-of-range percentages, were treated as missing because keeping impossible values would distort later analysis.
