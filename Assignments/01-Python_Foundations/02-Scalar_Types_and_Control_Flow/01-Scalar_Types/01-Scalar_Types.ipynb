# 📘 01 - Scalar Types

Review and practice working with Python's basic scalar data types.

## ✅ Learning Goals
- Recognize scalar types (`int`, `float`, `str`, `bool`)
- Use the `type()` function to identify types
- Use variables and assignment

````python
# 🔧 Define some variables
num = 42
pi = 3.14159
name = 'Python'
is_cool = True
````

````python
# 🔍 Check their types
print(type(num))
print(type(pi))
print(type(name))
print(type(is_cool))
````

### ✏️ Your Turn
Create your own variables of each type and print them.

````python
# Completed reference solution
count = 12          # int
price = 19.99       # float
course = "Spatial Data"  # str
is_ready = True     # bool

for value in [count, price, course, is_ready]:
    print(value, type(value))
````

````python
# Extra scalar examples
temperature = 72.5
city = "Wichita Falls"
has_data = True

print(temperature, type(temperature))
print(city, type(city))
print(has_data, type(has_data))
````

## ⚖️ Comparison Practice
Use comparison operators to check values.

````python
x = 10
y = 5
print(x > y)
print(x == y)
print(x != y)
````
