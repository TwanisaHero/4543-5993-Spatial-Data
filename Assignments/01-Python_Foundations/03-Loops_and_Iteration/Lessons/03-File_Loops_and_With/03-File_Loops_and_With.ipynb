# 📘 03 - File Loops and `with`

Use `with open(...)` to read files and process content.

## ✅ Learning Goals
- Open and read files using `with`
- Loop over lines in a file
- Use `.strip()` and `.split()` to process text

```python
# Sample file read
from pathlib import Path

sample = Path("sample.txt")
if not sample.exists():
    sample.write_text("alpha,beta,gamma\none,two,three,four\n", encoding="utf-8")

with open("sample.txt", "r", encoding="utf-8") as file:
    for line in file:
        print(line.strip())
```

```python
# 🔍 Process each line
with open('sample.txt', 'r') as f:
    for line in f:
        parts = line.strip().split(',')
        print('Parts:', parts)
```
