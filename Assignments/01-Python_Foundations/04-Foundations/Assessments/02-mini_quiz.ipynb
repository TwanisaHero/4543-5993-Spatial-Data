## 3. Create a simple plot.

```python
# Answer 3
import matplotlib.pyplot as plt

x = [0, 1, 2, 3, 4]
y = [value ** 2 for value in x]
plt.plot(x, y, marker="o")
plt.title("Squares")
plt.xlabel("x")
plt.ylabel("x squared")
plt.show()
```
